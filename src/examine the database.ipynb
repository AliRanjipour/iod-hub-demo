```python
# examine the duckdb database, econ_facts table and master_en and fa tables

import duckdb
path = "/Users/alirezaranjipour/Documents/Work/IOD/IOD_Data_Dash/IOD-d_d_last_version/data/duckdb/iod_dash.duckdb"
con = duckdb.connect(path, read_only=True)

facts = con.execute("SELECT * FROM econ_facts").fetchdf()
master_en = con.execute("SELECT * FROM master_en").fetchdf()
master_fa = con.execute("SELECT * FROM master_fa").fetchdf()
geo = con.execute("SELECT * FROM geo").fetchdf()
sub = con.execute("SELECT * FROM sub").fetchdf()

print(facts.columns.tolist())
print(master_en.columns.tolist())
print(master_fa.columns.tolist())
print(geo.columns.tolist())
print(sub.columns.tolist())

print(master_en[['topic_id', 'indicator_label']].head())

con.close()
```

```python

from doctest import master


df_fa = (
    facts[['topic_id','time_id','value','geo_code','sub_code', 'detail_fa']]
    .merge(
        master_fa[[ 'topic_id', 'indicator_label', 'default_unit', 'calendar', 'default_source']].drop_duplicates(subset='topic_id'),
        on="topic_id",
        how="left",
        validate="many_to_one"
    )
    .merge(
        geo[['geo_code', 'geo_label_fa']].drop_duplicates(subset='geo_code'),
        on="geo_code",
        how="left",
        validate="many_to_one"
    )
    .merge(
        sub[['sub_code', 'sub_label_fa']].drop_duplicates(subset='sub_code'),
        on="sub_code",
        how="left",
        validate="many_to_one"
    )
    .rename(columns={'geo_label_fa': 'geo_name', 'sub_label_fa': 'sub_label'})
)

df_fa = df_fa[
    ['indicator_label','geo_name', 'sub_label', 'time_id', 'value', 'default_unit', 'calendar', 'default_source', 'detail_fa']]
df_fa

```

```python
df_en = (
    facts[['topic_id','time_id','value','geo_code','sub_code', 'detail']]
    .merge(
        master_en[[ 'topic_id', 'indicator_label', 'default_unit', 'calendar', 'default_source']].drop_duplicates(subset='topic_id'),
        on="topic_id",
        how="left",
        validate="many_to_one"
    )
    .merge(
        geo[['geo_code', 'geo_label_en']].drop_duplicates(subset='geo_code'),
        on="geo_code",
        how="left",
        validate="many_to_one"
    )
    .merge(
        sub[['sub_code', 'sub_label_en']].drop_duplicates(subset='sub_code'),
        on="sub_code",
        how="left",
        validate="many_to_one"
    )
    .rename(columns={'geo_label_en': 'geo_name', 'sub_label_en': 'sub_label'})
)

df_en = df_en[
    ['indicator_label','geo_name', 'sub_label', 'time_id', 'value', 'default_unit', 'calendar', 'default_source', 'detail']]
df_en
```
